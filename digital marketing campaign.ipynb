```python
#import the necessary libraries
import pandas as pd
import numpy as np
```

```python
# load the dataset
data = pd.read_csv('global_ads_performance_dataset.csv')
#show first 5 rows of the dataset
data.head()
```

```python
#check for missing values
print(data.isnull().sum())
#check the data types of the columns 
data.dtypes
```

```python
#feature engineering
#date column to datetime format
data['date'] = pd.to_datetime(data['date'])
data['month'] = data['date'].dt.month
data['day'] = data['date'].dt.day
data['year'] = data['date'].dt.year
data['quarter'] = data['date'].dt.quarter  #create a new column for quarter of the year
#calculate profit
data['profit'] = data['revenue'] - data['ad_spend']
```

```python
# channel performance analysis
channel_performance = data.groupby('platform').agg(
    total_spend=('ad_spend', 'sum'),
    total_revenue=('revenue', 'sum'),
    total_profit=('profit', 'sum'),
    total_conversions=('conversions', 'sum'),
    total_clicks=('clicks', 'sum'),
    total_impressions=('impressions', 'sum'),
    avg_cpc=('CPC', 'mean'),
    avg_ctr=('CTR', 'mean'),
    avg_roas=('ROAS', 'mean'),
    avg_cpa=('CPA', 'mean')
).reset_index()
print(channel_performance)
```
